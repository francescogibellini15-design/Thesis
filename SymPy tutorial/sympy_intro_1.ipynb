```python
#Simbolic computation: roots are simplfied to rational factors
sp.sqrt(8)

#Variables are defined as symbols
x,y = sp.symbols('x y')

#Defining a symbolic expression
expr  = x + 2*y

#Can play around with expr
print(expr - x)

#Easy simplifications are handled but

print(x*expr)

#more comples ones are not automatic, one can use
expanded = sp.expand(x*expr)
print(expanded)
factor = sp.factor(expanded)
print(factor)
```

```python
x,t,z,nu = sp.symbols('x t z nu')

sp.init_printing(use_latex=True)
```

```python
#Derivatives
sp.diff(sp.sin(x)*sp.exp(x),x)

#Integrals
sp.integrate(sp.exp(x)*sp.sin(x) + sp.exp(x)*sp.cos(x), x)

#Integrals with limits, oo means infinity
sp.integrate(sp.sin(x**2), (x, -sp.oo, sp.oo))

#Limits
sp.limit(sp.sin(x)/x, x, 0)

#Equation solver
sp.solve(x**2 - 2, x)

#Solve differential equations

y = sp.Function('y')
sp.dsolve(sp.Eq(y(t).diff(t, t) - y(t), sp.exp(t)), y(t))

#.diff(t,t) is double derivative

#Find eigenvalues of a matrix
sp.Matrix([[1, 2], [2, 2]]).eigenvals()

#Rewrite bessel function as function of spherical bessel functions
sp.besselj(nu,z).rewrite(sp.jn)

#Print in latex 
sp.latex(sp.Integral(sp.cos(x)**2, (x,0, sp.pi)))
```

```python
#Must declare variables
x = symbols('x')
print(x+1)

x,y,z = symbols('x y z')

#Variable and symbol can be unrelated in name
a,b = symbols('c d')
print(a)

#Changing the variable doesn't affect the symbol
x = symbols('x')
expr = x + 1
print(expr)
x = 2
print(expr) #Unchanged

#But can substitute the values in
x = symbols('x')
expr = x+1
print(expr.subs(x,2))

#To use equality, use Eq(...)
Eq(x+1,4)

#Also for symbolic equality
(x+1)*2 == x**2 + 2*x + 1 #Gives false

#But can use simplidy, which isn't perfect but works for simple cases
a = (x+1)**2
b = x**2 + 2*x + 1
print(simplify(a-b))
c = x**2 - 2*x + 1
print(simplify(a-c))

# or can use .equals
a = cos(x)**2 - sin(x)**2
b = cos(2*x)
print(a.equals(b))
```

```python
#In python ^ is XOR

#Combining SymPy objects with SymPy or Python objects gives SymPy objects
#but two Python objects give Python objects.
#This affects division.
print(Integer(1)/Integer(2)) #Gives 1/2
#but
print(1/2) #gives 0.5

#so
print(x + 1/2) #gives x + 0.5

#to obtain + 1/2 use Rational
print(x + Rational(1,2))
```

```python
#BASIC OPERATIONS

#Substitution

expr = cos(x) + 1
expr.subs(x, y)

#Used for

expr.subs(x, 0) #Evaluation of expression at a point

#also to iterate expressions
expr = x**y
print(expr)

expr = expr.subs(y, expr)
print(expr)

expr = expr.subs(y, expr)
print(expr)

#Also to perform more advanced simplifications
expr = sin(2*x) + cos(2*x)
print(expand_trig(expr)) #expand_trig expands everything
print(expr.subs(sin(2*x), 2*sin(x)*cos(x))) #Subs can be more selective

#subs doesn't modify expr

#Can also do chain of subs
expr = x**3 +4*x*y - z
print(expr.subs([(x,2), (y,4), (z,0)]))

#Can also use list comprehension to do large set od subs altogether
expr = x**4 - 4*x**3 + 4*x**2 - 2*x + 3
replacements = [(x**i, y**i) for i in range(5) if i%2 == 0] #Changes all even powers of x to y
print(expr.subs(replacements))


```
