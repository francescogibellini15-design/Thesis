```python
#Printing functions
from sympy import *
init_printing()

#Converting to string
str(Integral(sqrt(1/x), x))
print(Integral(sqrt(1/x), x))

#Srepre gives exact expression form
print(srepr(Integral(sqrt(1/x), x)))

#ASCII pretty printer
pprint(Integral(sqrt(1/x), x), use_unicode = False)

#can also do in two steps
pretty(Integral(sqrt(1/x), x), use_unicode = False) #string
print(pretty(Integral(sqrt(1/x), x), use_unicode = False))

#UNICODE pretty printer
pprint(Integral(sqrt(1/x), x), use_unicode = True)

#LATEX
print(latex(Integral(sqrt(1/x), x)))

#latex has options to change formatting

#MathML
from sympy.printing.mathml import print_mathml 
print_mathml(Integral(sqrt(1/x),x))

#Dot
from sympy.printing.dot import dotprint
from sympy.abc import x

print(dotprint(x+2)) #can be rendered with Graphviz
```
